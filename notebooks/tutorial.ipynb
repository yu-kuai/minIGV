Uses the public ENCODE K562 snapshots in `examples/raw/` (see the README for the data).

```python
import sys, os, glob
sys.path.insert(0, '..')                 # repository root, where minigv.py lives
import importlib, minigv; importlib.reload(minigv)
from minigv import clean_svg, list_tracks, interactive

RAW = '../examples/raw'
```

## 1. Clean one snapshot
`clean_svg()` returns a figure; on the last line of a cell it previews inline.

```python
fig = clean_svg(os.path.join(RAW, 'HOXA_K562.svg'))
fig
```

```python
print(fig)          # what was removed / changed
```

## 2. Before / after

```python
fig.compare()
```

## 3. Options

```python
# also drop the coordinate ruler
clean_svg(os.path.join(RAW, 'MYC_K562.svg'), drop_ruler=True)
```

```python
# keep the ideogram and the span bar, use Helvetica
clean_svg(os.path.join(RAW, 'MYC_K562.svg'), keep_ideogram=True, keep_span=True, font='Helvetica')
```

```python
# rename tracks (substring replace); tidy_names=True would also drop .bw/.bam suffixes
clean_svg(os.path.join(RAW, 'MYC_K562.svg'), rename={'Pol II pS2': 'RNAPII Ser2P'})
```

```python
help(clean_svg)
```

## 4. Empty space
IGV pads each panel; by default blank bands taller than `min_gap` pt shrink to `gap` pt.

```python
p = os.path.join(RAW, 'GAPDH_K562.svg')
clean_svg(p, keep_gaps=True).plot(width='60%')   # IGV spacing
clean_svg(p).plot(width='60%')                   # gaps closed
```

## 5. Drop whole tracks

```python
list_tracks(os.path.join(RAW, 'GAPDH_K562.svg'))
```

```python
clean_svg(os.path.join(RAW, 'GAPDH_K562.svg'), drop_tracks=['H3K27me3', 'H3K27me3 peaks'])
```

## 6. Interactive
(Also in `interactive.ipynb`.)

```python
interactive(os.path.join(RAW, 'MYC_K562.svg'))
```

## 7. Batch with the same settings
Paste the call printed by the interactive tool into a loop.

```python
os.makedirs('cleaned', exist_ok=True)
for path in sorted(glob.glob(os.path.join(RAW, '*.svg'))):
    out = os.path.join('cleaned', os.path.basename(path).replace('.svg', '.clean.svg'))
    clean_svg(path, drop_ruler=True).save(out)
    print('saved', out)
```
